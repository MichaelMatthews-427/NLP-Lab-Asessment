```python
!pip install nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import nltk
nltk.download('punkt')
```

```python
text="this is a no. sentence"
tokens=word_tokenize(text)
print(tokens)
```

```python
import string 
text="! hey, i am *# ,"
tokenize=word_tokenize(text)
new_text=[]
for i in tokens:
    if i not in string.punctuation:
        new_text.append(i)
new_text
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
text="this is a no. sentence"
tokens=word_tokenize(text)
clean_text=[]
for i in tokens:
    if i.lower() not in stopwords.words("english"):
        clean_text.append(i)
clean_text
```

```python
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
words=["playing","played","unplayed","player","players","playful"]
for i in words:
    print(i,"->",stemming.stem(i.lower()))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemm=WordNetLemmatizer()
words=["fairly","easily","happy","happiness","unplayed","played"]
for i in words:
     print(lemm.lemmatize(i))
```
